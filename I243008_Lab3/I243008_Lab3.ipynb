```python
PEAS = {
    "Performance": ["delivery_time", "path_efficiency", "successful_deliveries"],
    "Environment": ["warehouse_grid", "storage_racks", "loading_dock"],
    "Actuators": ["move_up", "move_down", "move_left", "move_right"],
    "Sensors": ["position_sensor", "obstacle_sensor"]
}
print(PEAS)

env_type = {
    "observable": "full",    
    "deterministic": True,   
    "episodic": False,
    "static": True,          
    "discrete": True,
    "agents": 1
}
print(env_type)

class WarehouseRobotAgent:
    

    def __init__(self, goal_location):
        self.goal_location = goal_location  

    def goal_test(self, state):
        
        return state == self.goal_location

    def agent_program(self, state):
        
        if self.goal_test(state):
            return "Goal Reached - Stop"

        x, y = state
        gx, gy = self.goal_location

        
        if x < gx:
            return "move_right"
        elif x > gx:
            return "move_left"
        elif y < gy:
            return "move_up"
        elif y > gy:
            return "move_down"

    def apply_action(self, state, action):
      
        x, y = state
        if action == "move_right":
            x += 1
        elif action == "move_left":
            x -= 1
        elif action == "move_up":
            y += 1
        elif action == "move_down":
            y -= 1
        return (x, y)

loading_dock = (0, 0)
robot = WarehouseRobotAgent(goal_location=loading_dock)

start_positions = [(3, 2), (-2, 4), (1, -1)]

for start in start_positions:
    print(f"\nStarting position: {start}")
    state = start
    steps = 0
    while not robot.goal_test(state):
        action = robot.agent_program(state)
        print(f"  step {steps}: state={state} -> action={action}")
        state = robot.apply_action(state, action)
        steps += 1
        if steps > 20:  # safety limit to avoid infinite loops
            print("  Exceeded step limit, stopping.")
            break
    if robot.goal_test(state):
        print(f"  Goal reached at {state} in {steps} steps!")


```

```python
PEAS = {
    "Performance": ["patient_safety", "response_time", "false_alarm_rate"],
    "Environment": ["hospital_ward", "icu_bed", "patient_body"],
    "Actuators": ["display_alert", "notify_nurse", "notify_doctor"],
    "Sensors": ["heart_rate_sensor", "oxygen_sensor", "blood_pressure_sensor"]
}
print(PEAS)

env_type = {
    "observable": "partial",     
    "deterministic": True,    
    "episodic": False,        
    "static": False,         
    "discrete": True,
    "agents": 1
}
print(env_type)


class PatientMonitorAgent:

    def __init__(self):
        self.history = []

        self.normal_ranges = {
            "heart_rate": (60, 100),            
            "oxygen_saturation": (95, 100),   
            "blood_pressure_systolic": (90, 130)   
        }

    def is_abnormal(self, reading):
        hr_low, hr_high = self.normal_ranges["heart_rate"]
        o2_low, o2_high = self.normal_ranges["oxygen_saturation"]
        bp_low, bp_high = self.normal_ranges["blood_pressure_systolic"]

        return (
            not (hr_low <= reading["heart_rate"] <= hr_high)
            or not (o2_low <= reading["oxygen_saturation"] <= o2_high)
            or not (bp_low <= reading["blood_pressure_systolic"] <= bp_high)
        )

    def update_model(self, percept):
     
        self.history.append(percept)
        self.history = self.history[-5:]

    def agent_program(self, percept):
        
        self.update_model(percept)

        current_abnormal = self.is_abnormal(percept)
        abnormal_count = sum(self.is_abnormal(r) for r in self.history)

        if not current_abnormal:
            return "Continue Monitoring"
        elif abnormal_count >= 3:
            return "Call Doctor"
        else:
            return "Alert Nurse"

agent = PatientMonitorAgent()

readings = [
    {"heart_rate": 78,  "oxygen_saturation": 98, "blood_pressure_systolic": 120},
    {"heart_rate": 110, "oxygen_saturation": 93, "blood_pressure_systolic": 135},
    {"heart_rate": 115, "oxygen_saturation": 91, "blood_pressure_systolic": 140},
    {"heart_rate": 118, "oxygen_saturation": 90, "blood_pressure_systolic": 142},
    {"heart_rate": 80,  "oxygen_saturation": 97, "blood_pressure_systolic": 118},
]

for t, percept in enumerate(readings):
    action = agent.agent_program(percept)
    print(f"t={t} | percept={percept} -> action: {action}")
```
