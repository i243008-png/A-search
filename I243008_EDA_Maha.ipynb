```python
import pandas as pd
import matplotlib.pyplot as plt


house_df = pd.read_csv("house_prices.csv")


def iqr_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    
    print(f"\n=== {column} ===")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Lower Bound: {lower_bound}")
    print(f"Upper Bound: {upper_bound}")
    print(f"Number of Outliers: {outliers.shape[0]}")
    
    return outliers


outliers_saleprice = iqr_outliers(house_df, "SalePrice")
outliers_grlivarea = iqr_outliers(house_df, "GrLivArea")
outliers_garagearea = iqr_outliers(house_df, "GarageArea")


plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
house_df.boxplot(column="SalePrice")
plt.title("SalePrice Boxplot")

plt.subplot(1,3,2)
house_df.boxplot(column="GrLivArea")
plt.title("GrLivArea Boxplot")

plt.subplot(1,3,3)
house_df.boxplot(column="GarageArea")
plt.title("GarageArea Boxplot")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

bike_df = pd.read_csv("bike_sharing.csv")
house_df = pd.read_csv("house_prices.csv")

print("=== Bike-Sharing Demand: Working vs Non-Working Days ===")

bike_stats = bike_df.groupby("workingday")["casual"].agg(["mean", "median", "std", "min", "max"])
print(bike_stats)

if "registered" in bike_df.columns:
    bike_stats_all = bike_df.groupby("workingday")[["casual", "registered"]].agg(["mean", "median", "std"])
    print("\nCasual vs Registered Demand:")
    print(bike_stats_all)

print("\n=== House Prices Descriptive Statistics ===")

saleprice_stats = house_df["SalePrice"].agg(["mean", "median", "std", "min", "max"])
print("SalePrice Stats:")
print(saleprice_stats)

features = ["GrLivArea", "GarageArea", "OverallQual"]
feature_stats = house_df[features].describe()
print("\nOther Feature Stats:")
print(feature_stats)
```

```python
import seaborn as sns
import pandas as pd

penguins = sns.load_dataset("penguins")

print(penguins.head())
print(penguins.info())

print(penguins.isnull().sum())

num_cols = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]

for col in num_cols:
    penguins[col].fillna(penguins[col].median(), inplace=True)


penguins["sex"].fillna(penguins["sex"].mode()[0], inplace=True)

print(penguins.isnull().sum())


```
