```python
from functools import reduce

login_data = {
    "192.168.1.10": 3,
    "192.168.1.21": 17,
    "192.168.1.35": 5,
    "192.168.1.42": 24,
    "192.168.1.51": 11,
    "192.168.1.63": 2
}

suspicious_ips = list(filter(lambda item: item[1] > 10, login_data.items()))

risk_scores = list(map(lambda item: (item[0], item[1] * 2), suspicious_ips.items()))

total_risk_score = reduce(lambda acc, score: acc + score, risk_scores.values(), 0)

suspicious_set = set(suspicious_ips.keys())

highest_ip = max(suspicious_ips.items(), key=lambda item: item[1])

def classify_risk(attempts):
    if 11 <= attempts <= 15:
        return "Medium Risk"
    elif 16 <= attempts <= 20:
        return "High Risk"
    else:
        return "Critical Risk"

classified_ips = {ip: classify_risk(attempts) for ip, attempts in suspicious_ips.items()}

print("=== Security Report ===")
print("Suspicious IPs:", suspicious_set)
print("\nDetailed Risk Scores:")
for ip, score in risk_scores.items():
    print(f"{ip} -> Risk Score: {score}, Classification: {classified_ips[ip]}")

print(f"\nTotal Risk Score: {total_risk_score}")
print(f"Most Dangerous IP: {highest_ip[0]} with {highest_ip[1]} failed attempts ({classified_ips[highest_ip[0]]})")
```

```python
import numpy as np

books = np.array([
    [4.5, 120, 80],
    [4.2, 200, 65],
    [4.8, 150, 95],
    [3.9, 300, 55],
    [4.6, 180, 88]
])

print("Shape:", books.shape)   
print("Size:", books.size)     
print("Data type:", books.dtype)

column_means = np.mean(books, axis=0)
print("\nMean values (Rating, Views, Popularity):", column_means)

min_vals = books.min(axis=0)
max_vals = books.max(axis=0)
normalized_books = (books - min_vals) / (max_vals - min_vals)
print("\nNormalized data:\n", normalized_books)

scores = 0.5 * normalized_books[:, 0] + 0.2 * normalized_books[:, 1] + 0.3 * normalized_books[:, 2]
print("\nRecommendation Scores:", scores)

ranking = np.argsort(scores)[::-1]  
print("\nRanking (best to worst):", ranking)


print("\nBook Rankings and Scores:")
for idx in ranking:
    print(f"Book {idx+1}: Score = {scores[idx]:.4f}")


best_book = ranking[0]
print(f"\nBest Book: Book {best_book+1} with Score = {scores[best_book]:.4f}")

ratings = books[:, 0]
popularity = books[:, 2]
print("\nRatings:", ratings)
print("Popularity:", popularity)
```

```python
import numpy as np

field_data = [
    {"field": "F1", "temperature": 34, "humidity": 55, "soil_moisture": 28},
    {"field": "F2", "temperature": 29, "humidity": 72, "soil_moisture": 61},
    {"field": "F3", "temperature": 38, "humidity": 48, "soil_moisture": 22},
    {"field": "F4", "temperature": 31, "humidity": 65, "soil_moisture": 47},
    {"field": "F5", "temperature": 36, "humidity": 50, "soil_moisture": 30}
]

irrigation_fields = [f["field"] for f in field_data if f["soil_moisture"] < 30]

heat_stress_fields = [f["field"] for f in field_data if f["temperature"] > 35]

critical_fields = [f["field"] for f in field_data if f["soil_moisture"] < 30 and f["temperature"] > 35]

print("Fields requiring irrigation:", irrigation_fields)
print("Fields under heat stress:", heat_stress_fields)
print("Critical fields (both conditions):", critical_fields)

matrix = np.array([[f["temperature"], f["humidity"], f["soil_moisture"]] for f in field_data])
print("\nSensor Matrix:\n", matrix)

means = matrix.mean(axis=0)
print("\nMean Temperature:", means[0])
print("Mean Humidity:", means[1])
print("Mean Soil Moisture:", means[2])

variances = matrix.var(axis=0)
print("\nVariance (Temp, Humidity, Soil Moisture):", variances)

sensors = ["Temperature", "Humidity", "Soil Moisture"]
highest_var_sensor = sensors[np.argmax(variances)]
print("Sensor with highest variance:", highest_var_sensor)

min_vals = matrix.min(axis=0)
max_vals = matrix.max(axis=0)
normalized = (matrix - min_vals) / (max_vals - min_vals)

scores = 0.4 * normalized[:, 0] + 0.2 * normalized[:, 1] + 0.4 * (1 - normalized[:, 2])

ranking = np.argsort(scores)[::-1]

print("\n=== Field Risk Report ===")
for idx in ranking:
    field = field_data[idx]["field"]
    score = scores[idx]
    if score > 0.7:
        status = "High Risk → Immediate inspection required"
    elif score > 0.4:
        status = "Medium Risk → Monitor field"
    else:
        status = "Low Risk → Normal condition"
    print(f"{field}: Risk Score = {score:.3f}, {status}")
```

```python
import numpy as np
import math

stops = [
    ("Stop A", (33.6425, 73.1730)),
    ("Stop B", (33.6500, 73.1600)),
    ("Stop C", (33.6550, 73.1500)),
    ("Stop D", (33.6600, 73.1400)),
    ("Stop E", (33.6650, 73.1300))
]

stop_dict = {name: coords for name, coords in stops}

visited_stops = set()

def has_visited(stop):
    return stop in visited_stops

def visit_stop(stop):
    if not has_visited(stop):
        visited_stops.add(stop)

routes = [
    ["Stop A", "Stop B", ["Stop C", "Stop D"]],
    ["Stop A", ["Stop E"]]
]

def flatten_routes(route):
    flat = []
    for r in route:
        if isinstance(r, list):
            flat.extend(flatten_routes(r))  
        else:
            flat.append(r)
    return flat

flattened_route = flatten_routes(routes)

for stop in flattened_route:
    visit_stop(stop)

def calculate_distance(point1, point2):
    return math.sqrt((point1[0] - point2[0])**2 + (point1[1] - point2[1])**2)

coords_array = np.array([coords for _, coords in stops])

mean_lat = np.mean(coords_array[:, 0])
mean_lon = np.mean(coords_array[:, 1])
min_lat = np.min(coords_array[:, 0])
max_lat = np.max(coords_array[:, 0])

distances = []
for i in range(len(flattened_route) - 1):
    p1 = stop_dict[flattened_route[i]]
    p2 = stop_dict[flattened_route[i+1]]
    d = calculate_distance(p1, p2)
    distances.append((flattened_route[i], flattened_route[i+1], d))

closest_pair = min(distances, key=lambda x: x[2])

print("=== Intelligent Public Transport Route Analyzer ===")
print("\nFlattened Route:", flattened_route)
print("Visited Stops:", visited_stops)

print("\nRoute Statistics:")
print(f"Mean Latitude: {mean_lat:.4f}")
print(f"Mean Longitude: {mean_lon:.4f}")
print(f"Min Latitude: {min_lat:.4f}")
print(f"Max Latitude: {max_lat:.4f}")

print("\nDistances Between Consecutive Stops:")
for s1, s2, d in distances:
    print(f"{s1} → {s2}: {d:.4f}")

print(f"\nClosest Pair of Consecutive Stops: {closest_pair[0]} → {closest_pair[1]} ({closest_pair[2]:.4f})")
```
